```python
import pandas as pd
import numpy as np
```

1.Create a pandas DataFrame with 3 columns and 5 rows filled with random integers.Add a new column that is the product of the first two columns.

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']

print(df)
```

2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

```python
df = pd.DataFrame(np.random.randint(1, 10, (4, 3)),
                  columns=['A', 'B', 'C'])

print(df)

print("Row-wise sum:")
print(df.sum(axis=1))

print("Column-wise sum:")
print(df.sum(axis=0))
```

3. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Before filling:")
print(df)

df = df.fillna(df.mean())

print("After filling:")
print(df)
```

4. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan

print("Before dropping:")
print(df)

df = df.dropna()

print("After dropping:")
print(df)
```

5. Data Aggregation

Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(10, 100, 10)
})

print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print(result)
```

6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Scale', 'Box'],
    'Category': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Sales': np.random.randint(100, 1000, 6)
})

print(df)

result = df.groupby('Category')['Sales'].sum()

print(result)
```

7. Merging DataFrames

Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.



```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Ravi', 'John']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 90, 85]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

8. MultiIndex DataFrame

Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
index = pd.MultiIndex.from_tuples([
    ('A', 1),
    ('A', 2),
    ('B', 1),
    ('B', 2)
], names=['Category', 'Number'])

df = pd.DataFrame({
    'Value': [10, 20, 30, 40]
}, index=index)

print(df)

print("Category A:")
print(df.loc['A'])

print("Specific row:")
print(df.loc[('B', 1)])

print("Slicing:")
print(df.loc['A':'B'])
```

9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.



```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B', 'C', 'C'],
    'SubCategory': ['X', 'Y', 'X', 'Y', 'X', 'Y'],
    'Value': np.random.randint(10, 100, 6)
})

df = df.set_index(['Category', 'SubCategory'])

print(df)

print("Sum of values:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

10. Applying Functions

Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.



```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

result = df.apply(lambda x: x * 2)

print("After doubling:")
print(result)
```

11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.



```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 3)),
                  columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print(df)
```

12. Working with Text Data

Create a Pandas Series with 5 random text strings. Convert all the strings to uppercase.
